```python
# ==========================================
# Experiment Paths
# ==========================================
# Define all directories used in the Flask
# web application experiment.

from pathlib import Path

EXPERIMENT_DIR = Path("../experiments/E05_WebApplication")

FLASK_DIR = EXPERIMENT_DIR / "flask_app"

TEMPLATES_DIR = FLASK_DIR / "templates"

STATIC_DIR = FLASK_DIR / "static"

MODELS_DIR = FLASK_DIR / "models"

FLASK_DIR.mkdir(parents=True, exist_ok=True)

TEMPLATES_DIR.mkdir(parents=True, exist_ok=True)

STATIC_DIR.mkdir(parents=True, exist_ok=True)

MODELS_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# ==========================================
# Import Libraries
# ==========================================
# Import all libraries required by the
# Flask application.

import joblib

import numpy as np

import pandas as pd

from pathlib import Path
```

```python
# ==========================================
# Load Optimized Model
# ==========================================
# Load the optimized XGBoost model generated
# in Experiment 03.

model = joblib.load(

    "../experiments/E03_HyperparameterOptimization/trained_models/optimized_xgboost.pkl"

)

print("Optimized model loaded successfully.")
```

```python
# ==========================================
# Load Label Encoder
# ==========================================
# Load the label encoder used during training.

label_encoder = joblib.load(

    "../experiments/E03_HyperparameterOptimization/label_encoder.pkl"

)

print(label_encoder.classes_)
```

```python
# ==========================================
# Load Feature Names
# ==========================================
# Load the feature names required by the model.

feature_names = joblib.load(

    "../experiments/E03_HyperparameterOptimization/feature_names.pkl"

)

print(feature_names)

print()

print("Number of Features:", len(feature_names))
```

```python
# ==========================================
# Create Dummy Sample
# ==========================================
# Generate one dummy sample to verify that
# the model works correctly.

sample = pd.DataFrame(

    np.zeros((1, len(feature_names))),

    columns=feature_names

)

prediction = model.predict(sample)

print(prediction)
```

```python
# ==========================================
# Decode Prediction
# ==========================================
# Convert the predicted class index into
# its corresponding attack label.

attack_name = label_encoder.inverse_transform(

    prediction

)

print(attack_name)
```

```python
# ==========================================
# Copy Model Files
# ==========================================
# Copy all required files from E03 into
# the Flask application.

import shutil

source_dir = Path(

    "../experiments/E03_HyperparameterOptimization"

)

shutil.copy(

    source_dir / "trained_models" / "optimized_xgboost.pkl",

    MODELS_DIR / "optimized_xgboost.pkl"

)

shutil.copy(

    source_dir / "feature_names.pkl",

    MODELS_DIR / "feature_names.pkl"

)

shutil.copy(

    source_dir / "label_encoder.pkl",

    MODELS_DIR / "label_encoder.pkl"

)

print("All model files copied successfully.")
```
