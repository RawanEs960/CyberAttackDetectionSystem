```python
import pandas as pd

# قراءة الملف الأصلي
df = pd.read_csv("../data/cicids2017_cleaned.csv")

# =========================
# أخذ عينات من كل فئة
# =========================

normal_sample = df[df["Attack Type"] == "Normal Traffic"].sample(10, random_state=42)

dos_sample = df[df["Attack Type"] == "DoS"].sample(10, random_state=42)

ddos_sample = df[df["Attack Type"] == "DDoS"].sample(10, random_state=42)

portscan_sample = df[df["Attack Type"] == "Port Scanning"].sample(10, random_state=42)

bots_sample = df[df["Attack Type"] == "Bots"].sample(5, random_state=42)

web_sample = df[df["Attack Type"] == "Web Attacks"].sample(5, random_state=42)

# =========================
# دمج العينات
# =========================

demo_df = pd.concat([
    normal_sample,
    dos_sample,
    ddos_sample,
    portscan_sample,
    bots_sample,
    web_sample
])

# خلط الصفوف
demo_df = demo_df.sample(frac=1, random_state=42)

# =========================
# حفظ الملف
# =========================

demo_df.to_csv("../data/demo_attack.csv", index=False)

print("✅ Demo CSV created successfully!")
print("Shape:", demo_df.shape)

print("\nAttack Distribution:")
print(demo_df["Attack Type"].value_counts())
```
