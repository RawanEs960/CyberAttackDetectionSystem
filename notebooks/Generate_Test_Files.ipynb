```python
# ==========================================
# Generate Test Files for Flask Application
# ==========================================
# This notebook creates multiple CSV files
# from the CICIDS2017 dataset to test the
# Flask Web Application.
```

```python
# ==========================================
# Import Libraries
# ==========================================

import pandas as pd
from pathlib import Path
```

```python
# ==========================================
# Load Dataset
# ==========================================

DATA_PATH = "../data/cicids2017_cleaned.csv"

df = pd.read_csv(DATA_PATH)

print(df.shape)

df.head()
```

```python
# ==========================================
# Create Output Folder
# ==========================================

OUTPUT_DIR = Path("../test_files")

OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
# ==========================================
# Normal Traffic Only
# ==========================================

normal = (

    df[df["Attack Type"] == "Normal Traffic"]

    .sample(

        100,

        random_state=42

    )

)

normal.to_csv(

    OUTPUT_DIR /

    "normal_only.csv",

    index=False

)

print("normal_only.csv created")
```

```python
# ==========================================
# Mixed Traffic
# ==========================================

normal = df[df["Attack Type"] == "Normal Traffic"].sample(

    70,

    random_state=42

)

attack = df[df["Attack Type"] != "Normal Traffic"].sample(

    30,

    random_state=42

)

mixed = (

    pd.concat(

        [

            normal,

            attack

        ]

    )

    .sample(

        frac=1,

        random_state=42

    )

)

mixed.to_csv(

    OUTPUT_DIR /

    "mixed_traffic.csv",

    index=False

)

print("mixed_traffic.csv created")
```

```python
# ==========================================
# Attack Heavy
# ==========================================

normal = df[df["Attack Type"] == "Normal Traffic"].sample(

    10,

    random_state=42

)

attack = df[df["Attack Type"] != "Normal Traffic"].sample(

    90,

    random_state=42

)

attack_heavy = (

    pd.concat(

        [

            normal,

            attack

        ]

    )

    .sample(

        frac=1,

        random_state=42

    )

)

attack_heavy.to_csv(

    OUTPUT_DIR /

    "attack_heavy.csv",

    index=False

)

print("attack_heavy.csv created")
```

```python
# ==========================================
# Random Sample
# ==========================================

random_sample = df.sample(

    500,

    random_state=42

)

random_sample.to_csv(

    OUTPUT_DIR /

    "random_sample.csv",

    index=False

)

print("random_sample.csv created")
```

```python
# ==========================================
# Large Test File
# ==========================================

large = df.sample(

    5000,

    random_state=42

)

large.to_csv(

    OUTPUT_DIR /

    "large_test.csv",

    index=False

)

print("large_test.csv created")
```

```python
# ==========================================
# Missing Columns Test
# ==========================================
# Remove one important feature to test
# Flask validation.

missing = df.sample(

    100,

    random_state=42

).copy()

missing.drop(

    columns=["Flow Duration"],

    inplace=True

)

missing.to_csv(

    OUTPUT_DIR /

    "missing_columns.csv",

    index=False

)

print("missing_columns.csv created")
```

```python
# ==========================================
# Summary
# ==========================================

print("="*50)

print("Generated Test Files")

print("="*50)

for file in OUTPUT_DIR.glob("*.csv"):

    print(file.name)
```
